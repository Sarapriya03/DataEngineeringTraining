```python
from pyspark.sql.types import (
    StructType,
    StructField,
    IntegerType,
    StringType,
    DoubleType,
    BooleanType
)
```

```python
data = [
    (101, "Laptop", "Electronics", 65000.0, True),
    (102, "Mouse", "Accessories", 1500.0, True),
    (103, "Monitor", "Electronics", 18000.0, True),
    (104, "Office Chair", "Furniture", 9000.0, False),
    (105, "Keyboard", "Accessories", 3000.0, True)
]
```

```python
schema = StructType([
    StructField("product_id", IntegerType(), False),
    StructField("product_name", StringType(), False),
    StructField("category", StringType(), True),
    StructField("price", DoubleType(), True),
    StructField("available", BooleanType(), True)
])
```

```python
df = spark.createDataFrame(data, schema)
```

```python
df.show()
```

```python
df.printSchema()
```

```python
df.select("product_name","price","available").show()
```

```python
df.filter(df.available == False).show()
```

```python
from pyspark.sql.functions import col

df2 = df.withColumn("price with tax", col("price") * 1.18)
df2.show()
```

```python
from pyspark.sql.functions import col

df2 = df.withColumn(
    "price",
    col("price") * 1.18
)
df2.show()
```

```python
df2 = df.withColumnRenamed("product_name", "name")
df2.show()
```

```python
df2 = df.drop("available")
df2.show()
```

```python
from pyspark.sql.functions import lit

df2 = df.withColumn("country",lit("India"))
df2.show()
```

**Exercise Set 4 —
withColumn()**

Create:gross_amount using: quantity × unit_price

```python
from pyspark.sql.functions import col
```

23. Add the gross_amount column.


```python
df = df.withColumn("gross_amount",col("quantity") * col("unit_price"))
df.show()
```

24. Create: discount_amount using:
 gross_amount × discount_pct / 100

```python
df = df.withColumn("discount_amount",col("gross_amount") * col("discount_pct") / 100)
df.show()
```

25. Create: final_amount using:
 gross_amount - discount_amount

```python
df = df.withColumn("final_amount",col("gross_amount") - col("discount_amount"))
df.show()
```

26. Add a constant column: country = India

```python
df = df.withColumn("country",lit("India"))
df.show()
```

27. Add: high_value where:
 final_amount >= 50000 → Yes
otherwise             → No

```python
from pyspark.sql.functions import when

df = df.withColumn("high_value", when(col("final_amount") >= 50000,lit("Yes")).otherwise(lit("No")))
df.show()
```

```python
df.select("order_id","gross_amount","discount_amount","final_amount","country","high_value").show()
```

**Exercise Set 5 — Conditional Columns**

Create a column: order_size using:
 final_amount >= 50000 → High
final_amount >= 15000 → Medium
otherwise             → Low

```python
df = df.withColumn(
    "order_size",
    when(col("final_amount") >= 50000,lit("High")).when(col("final_amount") >= 15000,lit("Medium")).otherwise(lit("Low"))
)
```

28. Create order_size .

```python
df = df.withColumn(
    "order_size",
    when(col("final_amount") >= 50000,lit("High")).when(col("final_amount") >= 15000,lit("Medium")).otherwise(lit("Low"))
)
```

29. Create another column:
quantity_type where:
quantity >= 4 → Bulk
otherwise     → Regular

```python
df = df.withColumn(
    "quantity_type",
    when(col("quantity") >= 4,lit("Bulk")).otherwise(lit("Regular"))
)
```

30. Create a delivery-status category where:
Delivered          → Completed
Cancelled/Returned → Failed
Others             → In Progress
