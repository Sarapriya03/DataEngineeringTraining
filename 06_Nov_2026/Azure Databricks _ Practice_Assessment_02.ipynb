35. Explain the purpose of Parquet files and the _delta_log directory in a Delta table. Identify what information is stored in each.

**Parquet files** hold the actual table data: a compressed, columnar format that stores the row values column by column,along with the schema and per-column statistics (min/max, null counts) in each file footer. Delta never edits a Parquet file in place - an UPDATE/DELETE/MERGE writes new files and marks old ones as removed.

**`_delta_log`** is the transaction log. It holds ordered JSON commit files (`00000000000000000000.json`, ...) plus periodic Parquet checkpoint files. Each commit records the table's metadata (schema, partitioning, properties), which data files were **added** or **removed**, file-level statistics, and commit info (operation, timestamp, user). The current table state is the replay of this log. That is how Delta provides ACID transactions, time travel, and `DESCRIBE HISTORY`.

36. Explain what VACUUM does. Run VACUUM ... DRY RUN if supported by your environment, and identify which files would be eligible for removal without deleting them.

**`VACUUM`** permanently deletes data files that are no longer referenced by the current table version and are older than the retention threshold (default 7 days). It frees storage, but after it runs you can no longer time travel to versions that need the deleted files.
**`DRY RUN`** only lists the files that *would* be removed.
Dry run only - deletes nothing. With the default 7-day retention this is likely to return no rows, because the old files from your UPDATE/DELETE/MERGE steps were created today.

```python
%sql
VACUUM event_registrations_work DRY RUN;
```

**Scope and lifetime**

**Permanent view** (`event_revenue_view`): stored in the metastore/catalog; visible to any session or notebook with access; survives restarts until dropped.

**Temporary view**: exists only in the current Spark session (this notebook); disappears when the session ends or compute detaches.

**Global temporary view**: registered in the system `global_temp` schema and shared across notebooks/sessions on the same cluster/application,
queried as `global_temp.<name>`; disappears when the cluster/application terminates. Often unavailable on serverless.

**When ZORDER is useful:** Z-ordering co-locates rows with similar values of the chosen column(s) in the same files, which tightens each file's min/max statistics. Queries that filter on that column (e.g. `WHERE event_id = 'E02'`) can then skip most files (data skipping).It helps with large tables and high-cardinality columns that are frequently used in filters or joins.

46. Create a streaming DataFrame using Spark's built-in rate source, generating five records per second.

```python
rate_df = (spark.readStream
           .format("rate")
           .option("rowsPerSecond", 5)
           .load())
print(rate_df.isStreaming)
rate_df.printSchema()
```

47. Display the streaming DataFrame and identify its timestamp and value columns.

```python
display(rate_df, checkpointLocation="/Volumes/workspace/default/checkpoints/e_q47")
```

48. Use withColumn() to generate a transaction_id and a transaction_amount from the streaming values.

```python
from pyspark.sql.functions import col, concat, lit
 
txn_df = (rate_df
          .withColumn("transaction_id", concat(lit("TXN-"), col("value").cast("string")))
          .withColumn("transaction_amount", (col("value") % 10 + 1) * 100))   # 100 .. 1000
 
display(txn_df, checkpointLocation="/Volumes/workspace/default/checkpoints/e_q48")
```

49. Filter streaming records where transaction_amount exceeds ₹500.

```python
high_df = txn_df.filter(col("transaction_amount") > 500)
display(high_df, checkpointLocation="/Volumes/workspace/default/checkpoints/e_q49")
```

50. Aggregate streaming records by a derived transaction category, display the running counts using a supported streaming sink, and stop the query.

```python
from pyspark.sql.functions import when
 
cat_df = txn_df.withColumn(
    "transaction_category",
    when(col("transaction_amount") >= 800, "High")
    .when(col("transaction_amount") >= 400, "Medium")
    .otherwise("Low"))
 
counts_df = cat_df.groupBy("transaction_category").count()
```

```python
for s in spark.streams.active: s.stop()
```

```python
display(counts_df, checkpointLocation="/Volumes/workspace/default/checkpoints/e_q50")
```
