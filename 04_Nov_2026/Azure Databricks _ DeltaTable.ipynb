19. Display the complete transaction history using DESCRIBE HISTORY.

20. Find the table version before the updates and retrieve its records using Time Travel.

```python
%sql SELECT * FROM fitness_memberships_demo VERSION AS OF 2;
```

21. Compare the current records with the earlier version.

```python
%sql
SELECT 'Version 2' AS version, COUNT(*) AS total_rows
FROM fitness_memberships_demo VERSION AS OF 2
UNION ALL
SELECT 'Current', COUNT(*) FROM fitness_memberships_demo;
```
